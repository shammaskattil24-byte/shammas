```python

def fare_category(fare):
    if fare < 20:
        return "Low"
    elif fare <= 50:
        return "Medium"
    else:
        return "High"
df['Ticket_Fare_Category'] = df['Fare'].apply(fare_category)
print("\nFare Category")
print(df[['Fare', 'Ticket_Fare_Category']].head())
```
