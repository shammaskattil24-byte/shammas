```python
df.columns = [
    "employeeid",
    "employee_name",
    "age",
    "gender",
    "department",
    "salary",
    "city",
    "join_date",
    "performance",
    "email"
]
df
```

```python
object_cols = df.select_dtypes(include="object").columns

for col in object_cols:
    df[col]=df[col].astype(str).str.strip().str.title()
df
```

```python
df['age']=df['age'].replace("Thirty",30)
df
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df
```

```python
df["salary"] = df["salary"].astype(str)
df["salary"] = df["salary"].str.replace(",", "")
df["salary"] = df["salary"].replace("N/A", np.nan)
df["salary"] = pd.to_numeric(df["salary"])

df
```

```python
df["salary"] = df["salary"].fillna(df["salary"].median())
```

```python
df
```

```python
df["department"] = df["department"].fillna("Unknown")
df
```

```python
df["city"] = df["city"].fillna("Unknown")
df
```

```python
df["gender"] = df["gender"].replace("F","Female").replace("M","Male")
df["gender"] = df["gender"].fillna("Unknown")
df
```

```python
df["performance"] = df["performance"].fillna("Unknown")
df
```

```python
df["join_date"] = df["join_date"].str.replace("/", "-")
df
```

```python
df["join_date"] = df["join_date"].replace("N/A", np.nan)
df
```

```python
print(df["join_date"].head())
```

```python
df["join_date"] = pd.to_datetime(
    df["join_date"],
    format="mixed",
    errors="coerce"
)
df
```

```python
df["join_date"] = pd.to_datetime(df["join_date"])
df["join_date"] = df["join_date"].fillna(df["join_date"].mean())
df
```

```python
df["email"] = df["email"].fillna("Unknown")
df
```

```python
df = df.drop_duplicates(keep='first')
```
