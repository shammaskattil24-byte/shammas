```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel(r"C:\Users\Shamm\Downloads\c4fe3a02-800a-48ae-b5bc-15bbc219f46a (1).xlsx")
df
```

```python
plt.plot(df["Month"], df["Sales"], marker="o",color="green")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.pie(
    df["Profit"],
    labels=df["Month"],
    autopct="%1.1f%%"
)

plt.title("Monthly Profit Distribution")
plt.show()
```

```python
plt.scatter(df["Customers"], df["Sales"])

plt.title("Customers vs Sales")
plt.xlabel("Customers")
plt.ylabel("Sales")

plt.show()
```
